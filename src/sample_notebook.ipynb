## SDD Hello World (Minimalist Python)

Ejecución del programa Hello World bajo la Constitución SDD:
- Cero sobreingeniería (únicamente `print()`)
- Sin Spark, PySpark, DataFrames ni `display()`
- Validación estricta mediante aserciones nativas

```python
# Contrato de Verificación SDD (Fase 4)
mensaje = "Hello World"
assert isinstance(mensaje, str), "Violación de contrato: el mensaje debe ser de tipo str"
assert mensaje == "Hello World", "Violación de contrato: el contenido no coincide con 'Hello World'"
assert not hasattr(mensaje, "rdd"), "Violación de constitución: Objeto no debe ser una estructura Spark"
assert not hasattr(mensaje, "_jdf"), "Violación de constitución: Objeto contiene referencias JVM/Spark"

# Emisión canónica (Fase 3)
print(mensaje)
```
